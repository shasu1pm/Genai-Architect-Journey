# 2. Python & Its Uses

**Python** is a popular programming language created by **Guido van Rossum** in 1991.

It is famous because it is:
- **Easy to read** (looks almost like English)
- **Simple to learn** (great for beginners)
- **Powerful** (used by big companies like Google, Netflix, Instagram)

## Why is Python so popular?

Look at how simple it is. To show a message, we just write:

```python
print("Python is easy to learn!")

print("hello world")
```

```python
print("helo")
```

## Where is Python used? (Real uses)

| Field | Example |
|-------|---------|
| **Web Development** | Building websites (Instagram backend) |
| **Data Science** | Analyzing data, charts, statistics |
| **Artificial Intelligence (AI)** | ChatGPT, recommendation systems |
| **Machine Learning** | Predicting prices, spam detection |
| **Automation** | Doing boring tasks automatically |
| **Game Development** | Building simple games |
| **App Development** | Mobile and desktop apps |

## A small example of Python's power

With just a few lines, Python can do useful work — like calculating the average of marks.

```python
marks = [80, 90, 70, 100]
average = sum(marks) / len(marks)
print("Average marks:", average)
```

## Key points to remember

- Python is a beginner-friendly programming language.
- It is used in web, AI, data science, automation, and more.
- Its simple, English-like style makes it perfect to start with.
